```python
# 0.1 - Runtime check

import subprocess, shutil, os

print('---RAM---')
print(subprocess.run('free-g', shell = True, capture_output=True, text=True).stdout)
print('---CPU---')
print('cores:', os.cpu_count())
print('Disk content:')
print(f" total {shutil.disk_usage('/content').total/1e9:.0f} GB | free {shutil.disk_usage('/content').free/1e9:.0f} GB")

ram = int(subprocess.run('free -g |awk "NR==2{print \\$2}"', shell = True,
                         capture_output= True, text= True).stdout.strip() or 0)
print(f"\nRAM total: {ram} GB")
```
